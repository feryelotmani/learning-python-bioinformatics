```python
# Importing the SeqIO module from Biopython
# 'SeqIO' is used for reading and writing sequence files in a verity of formats
from Bio import SeqIO
```

```python
# this cell Reads a FASTA sequence file and converts it to an uppercase string
# 'SeqIO.read("/content/drive/MyDrive/HBB/sequence.fasta", "fasta")' reads a single sequence record from the imported FASTA file.
# 'record.seq' accesses the sequence object
# str(record.seq).upper() converts the seq object to a string and ensures all bases are uppercase for consistent processing
record=SeqIO.read("/content/drive/MyDrive/HBB/sequence.fasta", "fasta")
seq_str=str(record.seq).upper()
```

# **sequence** **analysis**
These cells perform basic analysis such as determining the length of the sequence and examines the specific regions





```python
# This cell Calculates and displays the length of the sequence
# 'len(seq_str)' returns the number of characters in the 'seq_str' string
length=(len(seq_str))
length
```

```python
# This cell Displays the first 80 characters of the sequence.
# Surving as a quick visual inspection of the sequence's beginning.
seq_str[:80]
```

```python
# This cell locates the starting position of the codon region
# 'seq_str.find("ATGGTGCATCTGACTCCTGA")' searches for the first occurrence of the given subsequence within 'seq_str' and returns its starting index
start_codon_pos=seq_str.find("ATGGTGCATCTGACTCCTGA")
start_codon_pos
```

```python
# This cell Extracts and displays the 7th codon after the identified start codon region
# 'start_codon_pos + 18' and 'start_codon_pos + 21' define the slice for the 7th codon
codon7=seq_str[start_codon_pos+18:start_codon_pos+21]
codon7
```

# **Simulating a Mutation (HBS - Sickle Cell Anemia)**
This section simulates the point mutation responsible for Sickle Cell Anemia (HbS) by changing a single nucleotide within a specific codon.

```python
# This cell simulates the HBS mutation by changing a nucleotide
# 'mut_pos = start_codon_pos + 19' identifies the exact position of the nucleotide to be changed within the 7th codon
# 'hbs_seq_str = seq_str[:mut_pos] + "T" + seq_str[mut_pos+1:]' creates a new sequence string (`hbs_seq_str`) where the nucleotide at 'mut_pos' is replaced with 'T'
mut_pos=start_codon_pos+19
hbs_seq_str=seq_str[:mut_pos]+"T"+ seq_str[mut_pos+1]
```

```python
# This cell Extracts and displays the mutated 7th codon
# This shows the specific codon after the simulated mutation
codon7_hbs=hbs_seq_str[start_codon_pos+18:start_codon_pos+21]
codon7_hbs
```

# **Translation**
This final section translates the mutated codon into its corresponding amino acid using Biopython's Seq object.

```python
# This cell Imports the 'Seq' object from Biopython
# The 'Seq' object is used to represent biological sequences and provides methods for various operations, including translation.
from Bio.Seq import Seq
```

```python
# This cell translates the mutated codon into an amino acid.
# 'Seq(codon7_hbs)' creates a 'Seq' object from the mutated codon string
# The function '.translate()' translates this DNA codon into its corresponding amino acid. In the case of the HbS mutation, the 'GAG' codon (Glutamic Acid) is changed to 'GTG' (Valine), which is represented as 'V'
codon7_hbs_translate=Seq(codon7_hbs).translate()
codon7_hbs_translate
```
